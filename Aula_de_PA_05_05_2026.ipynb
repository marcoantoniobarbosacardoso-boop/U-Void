```python
# 2  - Escrever um programa  que o usuário digita um número e o programa imprime número digitado até o zero. (contagem regressiva)

Número = int (input("Digite um número 0 a infinito: "))

for i in range (0, Número + 1):
  print (10 - i)
```

```python
# 3 - Escrever um programa que o usuário digita um número e o programa imprime os números pares.

Número = int (input("Digite um número aleatório: "))

for i in range (0, Número + 1):
  if i % 2 == 0 :
    print (i)
```

```python
# 4 - Escrever um programa que o usuário digita um número e o programa imprime os número ímpares.

Número = int (input("Digite um número: "))

for i in range (1, Número + 1, 2):
  print (i)
```

```python
# 5 - Escrever um programa que o usuário digita um número e o programa soma os números até o número digtitado.

numero = int(input("Digite um número inteiro positivo: "))

soma = 0

for i in range(1, numero + 1):
    soma += i

print(f"A soma de 1 até {numero} é: {soma}")
```
