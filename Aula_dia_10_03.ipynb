```python
# Entrada

Lado1 = float(input("Digite o comprimento"))
Lado2 = float(input("Digite a altura"))

# Processamento

area = Lado1 * Lado2 + Lado2 - Lado2 * Lado1

# Saída

print (f"O total da área é :{area} m²")
```
