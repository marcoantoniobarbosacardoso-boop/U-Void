```python
from pathlib import PureWindowsPath
#Operações Aritméticas

#Pedir ao usuário para digitar 2 números


#Entrada

a = int(input("Digite um número: "))
b = int(input("Digite um número: "))

#Processamento

#Soma

Soma = a + b
Subtração = a - b
Multiplicação = a * b
Divisão = a / b
Div_int = a // b
Resto = a % b
Potência = a ** b

#Saída

print("")
print("------------OPERADORES ARITMÉTICOS------------")
print("")
print("----------------------SOMA----------------------")
print("")
print(f"O resultado da DIFERENÇA de {a} MAIS {b} é igual a: {Soma}")
print("")
print("--------------------SUBTRAÇÃO-------------------")
print("")
print(f"O resultado da DIFERENÇA de {a} MENOS {b} é igual a: {Subtração}")
print("")
print("------------------MULTIPLICAÇÃO-----------------")
print("")
print(f"O resultado da DIFERENÇA DE {a} MULTIPLICADO por {b} é igual a: {Multiplicação}")
print("")
print("---------------------DIVISÃO--------------------")
print("")
print(f"O resultado da DIFERENÇA de {a} DIVIDIDO por {b} é igual a: {Divisão}")
print("")
print("---------------------DIV_INT--------------------")
print("")
print(f"O resultado da DIFERENÇA de {a} DIVIDO² por {b} é igual a: {Div_int}")
print("")
print("----------------------RESTO---------------------")
print("")
print(f"O resultado da DIFERENÇA de {a} RESTO {b} é igual a: {Resto}")
print("")
print("---------------------POTÊNCIA-------------------")
print("")
print(f"O resultado da DIFERENÇA de {a} POTÊNCIA {b} é igual a: {Potência}")
print("")
print("------------------------------------------------")
```
