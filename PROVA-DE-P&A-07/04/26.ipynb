```python
# Questão 1 :

idade1 = int (input ("Digite sua idade:"))


if idade1 < 17:
  print (f"Você é de menor.")

elif idade1 == 17:
  print (f"Você é de menor.")

elif idade1 < 59:
  print (f"Você é um adulto.")

elif idade1 > 59:
  print (f"Você é um idoso.")
```

```python
# Questão 2 :

número1 = int (input ("Digite um número: "))


if número1 < 100:
  print (f"Sem Desconto.")

elif número1 == 100:
  print (f"Sem Desconto.")

elif número1 < 500:
  print (número1 * 0.10)

elif número1 == 500:
  print (número1 * 0.10)

elif número1 > 500:
  print (número1 * 0.20)
```
