```python
palavra = str (input ("Digite uma palavra ou frase: "))
invertida = ""

for i in (palavra):

  invertida = i + invertida
print (invertida)
```

```python
Número = int(input("Digite um número de 1 a 10: "))

if 1 <= Número <= 10:
  print(f"Tabuada do {Número}:")
  for i in range(1, 11):

    print(f"{Número} x {i} = {Número * i}")
elif Número < 1 or Número > 10:
  print("Por favor, digite um número entre 1 e 10.")
```
